```python
df['dep_time'] = pd.to_datetime(
    df['dep_time'].astype('Int64').astype(str).str.zfill(4),
    format='%H%M',
    errors='coerce'
)
```

```python
df['dep_time'] = df['dep_time'].dt.strftime('%I:%M %p')
```

Note: For the above value after changing the data type of dep_time to hour minutes format, I get 82 null values because it is given as 24.00, which is an invalid time..

```python
# Creating a new column to analyse air_time

df['calculated_air_time'] = (((df['wheels_on'] // 100) * 60 + (df['wheels_on'] % 100))
    - ((df['wheels_off'] // 100) * 60 + (df['wheels_off'] % 100)))
```

```python
df
```

```python
df[df['air_time'] != df['calculated_air_time']]
```

Dropping some columns with more inconsistent values that may affect the accuracy.

```python
df = df.drop(columns=['wheels_on','wheels_off','air_time','calculated_air_time'],
    errors='ignore')
```

```python
df
```

Filling the null values in 'taxi_in' column using groupby and median

```python
df['taxi_in'] = df['taxi_in'].fillna(
    df.groupby('origin')['taxi_in'].transform('median')
)
```

```python
df['taxi_in'].isnull().sum()
```

```python
# Removed leading and trailling spaces using strip.
df.columns = df.columns.str.strip()
```

```python
df.skew(numeric_only=True)
```

```python
# To reduce skewness transform function using log1p has been used.

df['weather_delay_log'] = np.log1p(df['weather_delay'])
```

```python
df['weather_delay_log'].skew()
```

```python
df
```

Inserting a new day column from fl_date column with positioning next to fl_date column

```python
df.insert(
    df.columns.get_loc('fl_date') + 1,
    'day_name',
    df['fl_date'].dt.day_name()
)
```

```python
df
```

```python
df.dtypes
```

Inserting a new month_name column and stored it in the 6th index

```python
df['month_name'] = df['fl_date'].dt.month_name()
clm = df.pop('month_name')
df.insert(6, 'month_name', clm)
```

```python
df
```

```python
# Creating a new column 'ground_time'

df['ground_time'] = df['taxi_out'] + df['taxi_in']
df
```

```python
# Renaming the column names 'origin_city_name' and 'origin_state_nm'
df.rename(columns = {'origin_city_name': 'origin_city',
                     'origin_state_nm' : 'origin_state'}, inplace = True)
df
```

```python
q1 = df['ground_time'].quantile(0.33)
q2 = df['ground_time'].quantile(0.66)

df['congestion_level'] = pd.cut(df['ground_time'],
    bins=[-1, q1, q2, df['ground_time'].max()],
    labels=['Low', 'Medium', 'High'])

df
```

```python
df.dtypes
```

```python
def time_slot(hour):
    if 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'

df['time_slot'] = pd.to_datetime(df['dep_time']).dt.hour.apply(time_slot)
```

```python
df
```

```python
# dropping null values in dep_time.
df = df.dropna(subset=['dep_time'])
```
